```python
import torch

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM
)

model_name = "Helsinki-NLP/opus-mt-en-inc"

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Load AI model
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Select device
device = "cuda" if torch.cuda.is_available() else "cpu"

model = model.to(device)

print("AI Translation Model Loaded Successfully!")
print("Running on:", device)
```

```python
def translate_text(text, target_language="sin"):

    # Add target language code
    input_text = f">>{target_language}<< {text}"

    # Convert text into tokens
    inputs = tokenizer(
        input_text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    # Generate translation
    with torch.no_grad():
        translated = model.generate(
            **inputs,
            max_new_tokens=200
        )

    # Convert tokens into readable text
    result = tokenizer.decode(
        translated[0],
        skip_special_tokens=True
    )

    return result
```

```python
english_text = "Hello, how are you?"

sinhala_translation = translate_text(
    english_text,
    target_language="sin"
)

print("English:", english_text)
print("Sinhala:", sinhala_translation)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

text_input = widgets.Textarea(
    placeholder="Enter English text here...",
    description="English:",
    layout=widgets.Layout(width="80%", height="100px")
)

translate_button = widgets.Button(
    description="Translate to Sinhala",
    button_style="success"
)

output = widgets.Output()

def translate_button_clicked(b):

    with output:
        clear_output()

        if not text_input.value.strip():
            print("Please enter some text!")
            return

        result = translate_text(text_input.value)

        print("Sinhala Translation:")
        print(result)

translate_button.on_click(translate_button_clicked)

display(text_input, translate_button, output)
```
